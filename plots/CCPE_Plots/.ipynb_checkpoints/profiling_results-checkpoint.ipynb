```python
gid_map = {
    'profiling_agg': '1092807689'
}

sheet_id = "1e_lKct9ovnYByYkGpFCKrMpKQs5TQdKhTld9tDU1JcQ"
```

```python
import gdown
import pandas as pd
import numpy as np
from scipy import stats



def get_pandas_from_gid(gid_name):
    # Replace with your actual Google Sheet ID
    # Optionally: specify the sheet/tab number (gid=0 for the first sheet)
    gid = gid_map[gid_name]
    
    # Construct export URL
    url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=csv&gid={gid}"
    
    # Set desired output filename
    output = f"{gid_name}.csv"
    
    # Download CSV using gdown
    gdown.download(url, output, quiet=False)
    original_df = pd.read_csv(output)
    return original_df
```

```python
df = get_pandas_from_gid('profiling_agg')
df
```

```python
# Apply replacements globally
df['obj_function'] = df['obj_function'].replace({
    'conductance': 'CO',
    'triangledensestsubgraph': 'TDS',
    'densesubgraph': 'DS',
    'degreeentropy': 'DE',
    'labelentropy': 'LE'
})

df['graph'] = df['graph'].replace({
    'citeseer': 'Citeseer',
    'amazon': 'Amazon',
    'dblp': 'DBLP',
    'patents': 'Patents',
    'livejournal': 'LiveJournal',
    'youtube': 'Youtube',
})

df['metaheuristic'] = df['metaheuristic'].replace({
    'vns': 'VNS',
    'ils': 'ILS',
    'ts1': 'TS',
})

# Define the columns mapping
new_names = {
    'obj_function': 'Scoring Function',
    'graph': 'Graph',
    'AVERAGE of obj_function_perc': 'Time in Scoring Function (%)',
    'confidence_interval': 'Confidence Interval'
}

# Define sort orders
graph_order = ['Citeseer', 'Amazon', 'DBLP', 'Patents', 'LiveJournal', 'Youtube']
scoring_function_order = ['DS', 'CO',  'DE', 'LE', 'TDS']

# Loop through each metaheuristic to create separate tables
target_metaheuristics = ['VNS', 'ILS', 'TS']
for mh in target_metaheuristics:
    print(f"%% --- Table for {mh} ---")

    # Filter data for the current metaheuristic
    df_subset = df[df['metaheuristic'] == mh].copy()

    # Select and rename columns
    df_latex = df_subset[list(new_names.keys())].rename(columns=new_names).copy()

    # Format numbers to 2 decimals BEFORE converting to string
    # (Because to_latex float_format ignores string columns)
    df_latex['Mean'] = df_latex['Mean'].apply(lambda x: f"{x:.2f}")
    df_latex['CI'] = df_latex['CI'].apply(lambda x: f"{x:.2f}")

    # Combine Mean and CI with the latex symbol
    df_latex['Time in Scoring Function (%)'] = (
            df_latex['Mean'] + r" $\pm$ " + df_latex['CI']
    )

    # Drop the temporary separate columns
    df_latex = df_latex.drop(columns=['Mean', 'CI'])

    # Set Categorical Order
    df_latex['Graph'] = pd.Categorical(
        df_latex['Graph'], categories=graph_order, ordered=True
    )
    df_latex['Scoring Function'] = pd.Categorical(
        df_latex['Scoring Function'], categories=scoring_function_order, ordered=True
    )

    # Sort
    df_latex = df_latex.sort_values(by=['Scoring Function', 'Graph'])

    # Generate Latex
    # Note: We pivot or structured it to be cleaner?
    # Your original script kept it long-format. If you want a pivot (Graphs as columns),
    # let me know. Keeping it as requested:

    caption_text = f"Profiling of Scoring Functions - {mh}"
    label_text = f"tab:scoring-functions-{mh.lower()}"

    table_str = df_latex.to_latex(
        caption=caption_text,
        label=label_text,
        index=False,
        column_format="llc", # l=left align (text), c=center (numbers)
        escape=False # Important to render the $\pm$ correctly
    )

    print(table_str)
    print("\n")
```
